## 2. Como os arquivos são obtidos e conferidos

A Prefeitura publicou os dois CSVs com convenções distintas. O arquivo de 2025 usa ponto e vírgula como separador e dia/mês/ano nas datas. O de 2026 usa vírgula e mês/dia/ano. Aplicar a configuração correspondente a cada período evita trocar dia e mês ou ler uma linha inteira como se fosse uma única coluna.

A assinatura SHA-256 é um valor calculado a partir do conteúdo completo de cada arquivo. Durante a transferência, o Colab guarda os blocos recebidos em um arquivo temporário de extensão `.part`. A mensagem **download verificado** indica que a assinatura coincidiu com a da edição estudada. Em outra execução na mesma sessão, **cópia local verificada** indica que o arquivo já pode ser lido.

```python
from concurrent.futures import ThreadPoolExecutor
from pathlib import Path
from hashlib import sha256
from tempfile import NamedTemporaryFile
from urllib.request import Request, urlopen
from urllib.error import URLError
from http.client import HTTPException
import pandas as pd
import matplotlib.pyplot as plt

fontes = [{'periodo': '2025-T4',
  'url': 'https://dados.prefeitura.sp.gov.br/dataset/0aecfa2b-aa3a-40d4-8183-0d4351b7fd0a/resource/d3a98097-8128-45d4-b8fe-21874b81c971/download/arquivo-final-4-trim-25.csv',
  'arquivo': 'arquivo-final-4-trim-25.csv',
  'encoding': 'cp1252',
  'sep': ';',
  'data': '%d/%m/%Y',
  'sha256': '0a19e3451e30f33f1a8c2581608e1cc92d036915d0594737f0a2ceb0e05929cc'},
 {'periodo': '2026-T3',
  'url': 'https://dados.prefeitura.sp.gov.br/dataset/0aecfa2b-aa3a-40d4-8183-0d4351b7fd0a/resource/f67fcc12-8da4-485f-92a3-d2af40cc771e/download/arquivo-final-3-trim-26.csv',
  'arquivo': 'arquivo-final-3-trim-26.csv',
  'encoding': 'utf-8',
  'sep': ',',
  'data': '%m/%d/%Y',
  'sha256': '4e7eb68356c64178dc326feef3f65aeadb34f5507d484f13f3ea4b6bba341634'}]
```

```python
def assinatura(caminho):
    resumo = sha256()
    with caminho.open("rb") as entrada:
        for bloco in iter(lambda: entrada.read(1024 * 1024), b""):
            resumo.update(bloco)
    return resumo.hexdigest()


def obter_csv(fonte):
    caminho = Path(fonte["arquivo"])
    if caminho.is_file():
        if assinatura(caminho) == fonte["sha256"]:
            print(f"{caminho}: cópia local verificada.", flush=True)
            return caminho
        print(f"{caminho}: cópia divergente. Obtendo a edição auditada.", flush=True)

    for tentativa in range(1, 3):
        print(f"{caminho}: iniciando download ({tentativa}/2).", flush=True)
        with NamedTemporaryFile(dir=caminho.parent, suffix=".part", delete=False) as temporario:
            parcial = Path(temporario.name)
        try:
            recebido = 0
            proximo_aviso = 10_000_000
            pedido = Request(fonte["url"], headers={"User-Agent": "Polieus-Colab/1.0"})
            with urlopen(pedido, timeout=90) as resposta, parcial.open("wb") as saida:
                while bloco := resposta.read(1024 * 1024):
                    saida.write(bloco)
                    recebido += len(bloco)
                    if recebido >= proximo_aviso:
                        print(f"  {caminho}: {recebido / 1_000_000:.0f} MB recebidos", flush=True)
                        proximo_aviso += 10_000_000
            if assinatura(parcial) != fonte["sha256"]:
                raise ValueError("O conteúdo recebido diverge da edição auditada.")
            parcial.replace(caminho)
            print(f"{caminho}: download verificado.", flush=True)
            return caminho
        except (URLError, HTTPException, OSError, ValueError) as erro:
            print(f"Tentativa {tentativa}: {erro}", flush=True)
            if tentativa == 2:
                raise RuntimeError(
                    f"Confira o acesso ao portal ou envie uma cópia auditada com o nome {caminho}. "
                    "Se a Prefeitura republicou os dados, confira a edição antes de refazer a análise."
                ) from erro
        finally:
            parcial.unlink(missing_ok=True)


# Cópias de versões anteriores do projeto também podem ser reutilizadas.
for fonte in fontes:
    ano, trimestre = fonte["periodo"].split("-T")
    antigo = Path(f"SP156_Solicitacoes_{ano}_Trimestre_{trimestre}.csv")
    if not Path(fonte["arquivo"]).exists() and antigo.is_file() and assinatura(antigo) == fonte["sha256"]:
        fonte["arquivo"] = antigo.name

with ThreadPoolExecutor(max_workers=2) as executor:
    arquivos_verificados = list(executor.map(obter_csv, fontes))
```

## 3. Quais solicitações entram no estudo?

O recorte usa o nome completo do serviço escolhido no SP156. Para pavimentação, distingue tapa-buraco em vias comuns e em corredores ou faixas exclusivas de ônibus. Para circulação de pedestres, inclui avaliação de obstáculos, manutenção de calçada pública, denúncia de calçada particular danificada ou inexistente e rebaixamento de guias. Selecionar apenas pela palavra “calçada” poderia incluir solicitações sobre outros assuntos, como árvores.

A célula lê cinco campos: serviço, distrito, status, data de abertura e data de finalização. Em seguida, apresenta o total de linhas de cada trimestre e a quantidade que corresponde aos oito serviços. A última tabela mostra como essa seleção se distribui entre os serviços. O endereço detalhado é dispensável para comparar distritos e permanece fora das tabelas preparadas.

Os dois serviços complementares entram pelo nome oficial. Um pedido sobre horário de ônibus tem outro significado e fica fora da seleção.

### Os oito serviços utilizados

A seleção usa o nome registrado na fonte. Os exemplos ilustram situações de uso. Cada chamado entra uma vez na contagem de seu serviço.

1. **Tapa-buraco.** Pedido de reparo localizado no pavimento de uma via. Um exemplo é um buraco na pista de tráfego comum. O SP156 distingue os pedidos associados a corredores e faixas exclusivas de ônibus.

2. **Tapa-buraco em faixa exclusiva ou corredor de ônibus.** Pedido de reparo no pavimento de uma faixa exclusiva ou corredor de ônibus. A localização operacional distingue essa classe do tapa-buraco em via comum.

3. **Acessibilidade - solicitar avaliação de obstáculo na calçada.** Solicitação de avaliação de uma barreira no passeio que dificulta a circulação. O registro representa um pedido de vistoria ou avaliação. A remoção efetiva do obstáculo exige confirmação posterior.

4. **Calçada pública - solicitar manutenção.** Pedido de manutenção de uma calçada identificada como pública. Um exemplo é um trecho de passeio público com piso danificado que dificulta a passagem a pé.

5. **Calçada particular - denunciar calçada danificada ou inexistente.** Denúncia de passeio particular danificado ou inexistente. O trâmite pode envolver fiscalização. A duração administrativa precisa ser interpretada conforme esse tipo de atendimento.

6. **Guias para travessia de pedestres - Solicitar rebaixamento.** Pedido de rebaixamento de guia em travessia de pedestres, por exemplo para facilitar a passagem de cadeira de rodas entre a calçada e a pista.

7. **Solicitar limpeza de bueiros, boca de lobo e poços de visita.** Solicitação de limpeza de estruturas da drenagem urbana, como uma boca de lobo obstruída. A categoria permite observar demandas de escoamento de água em vias de circulação.

8. **Ponto de ônibus – Solicitar manutenção, conserto ou limpeza de equipamento existente.** Pedido de manutenção, conserto ou limpeza de equipamento já existente em ponto de ônibus. Um exemplo é um abrigo danificado. Reclamações sobre intervalos de linhas e pedidos de implantação têm outras categorias.

```python
servicos = ['Tapa-buraco', 'Tapa-buraco em faixa exclusiva ou corredor de ônibus', 'Acessibilidade - solicitar avaliação de obstáculo na calçada', 'Calçada pública - solicitar manutenção', 'Calçada particular - denunciar calçada danificada ou inexistente', 'Guias para travessia de pedestres - Solicitar rebaixamento']
servicos.extend(['Solicitar limpeza de bueiros, boca de lobo e poços de visita', 'Ponto de ônibus – Solicitar manutenção, conserto ou limpeza de equipamento existente'])
partes = []
for fonte in fontes:
    arquivo = Path(fonte['arquivo'])
    cabecalho = pd.read_csv(arquivo, encoding=fonte['encoding'], sep=fonte['sep'], nrows=0)
    campos = ['Serviço', 'Distrito', 'Status', 'Data de Abertura', 'Data de Finalização']
    dados = pd.read_csv(arquivo, encoding=fonte['encoding'], sep=fonte['sep'], usecols=campos, dtype=str)
    # A seleção exata preserva o mesmo critério nos dois períodos.
    recorte = dados.loc[dados['Serviço'].isin(servicos)].copy()
    recorte['periodo'] = fonte['periodo']
    for campo, destino in [('Data de Abertura', 'abertura'), ('Data de Finalização', 'finalizacao')]:
        recorte[destino] = pd.to_datetime(recorte[campo], format=fonte['data'], errors='coerce')
        invalidos = recorte[campo].notna() & recorte[destino].isna()
        print(fonte['periodo'], campo, 'preenchidos com formato inválido:', invalidos.sum())
    print(fonte['periodo'], 'linhas:', len(dados), 'selecionadas:', len(recorte))
    print('Cabeçalho:', list(cabecalho))
    partes.append(recorte)
df = pd.concat(partes, ignore_index=True)
df['dias'] = (df['finalizacao'] - df['abertura']).dt.days
df['mes'] = df['abertura'].dt.to_period('M')
print(pd.crosstab(df['Serviço'], df['periodo']).to_string())
```

## 4. O que as datas e o território permitem observar?

Primeiro conferimos se as datas preenchidas puderam ser interpretadas. Depois calculamos os dias entre abertura e finalização. Um chamado aberto e finalizado no mesmo dia tem duração de zero dias. Uma data final vazia deixa a duração desconhecida. Um intervalo negativo indicaria que a finalização registrada antecede a abertura e exigiria investigação.

A tabela de status mostra quantos registros têm data final em cada situação administrativa. As medianas desta seção consideram os intervalos calculáveis antes da seleção definitiva do alvo. Para o território, a saída informa quantos registros estão sem distrito e quantos pares de distrito e mês têm poucas solicitações. Esses números orientam a interpretação das duas tabelas seguintes.

```python
for periodo, dados in df.groupby('periodo'):
    perfis = dados.groupby(['Distrito', 'mes']).size()
    validos = dados['dias'].ge(0)
    print(periodo, 'aberturas:', dados.abertura.min(), dados.abertura.max())
    print('Distrito ausente:', dados.Distrito.isna().sum(), 'nomes distintos:', dados.Distrito.nunique())
    print('Perfis distrito-mês:', len(perfis), 'com menos de 5 registros:', (perfis < 5).sum())
    print('Intervalos não negativos, antes do filtro de status:', validos.sum(), 'negativa:', dados.dias.lt(0).sum())
    print(pd.crosstab(dados.Status.fillna('(ausente)'), dados.finalizacao.notna()).rename(columns={False: 'Sem data final', True: 'Com data final'}).to_string())
    print(dados.loc[validos].groupby('Serviço').dias.agg(['count', 'median']).rename(columns={'count': 'Registros', 'median': 'Mediana (dias)'}).to_string())
```

**Como ler a tabela de datas.** Cada linha representa um status. As duas colunas contam os pedidos com e sem data de finalização. Um pedido cancelado pode ter data de finalização. A próxima etapa aplica também o critério de status Finalizada para formar a amostra de regressão. A mediana indica o valor central das durações observadas em cada serviço.

## 5. Como se forma a amostra para regressão?

A pergunta é quanto tempo decorre da abertura até a finalização administrativa. Selecionamos solicitações com status **Finalizada** e duração válida em dias corridos. Cancelamentos possuem significado próprio e ficam fora dessa amostra. O total inicial do serviço continua registrado para que o leitor veja a parcela que pôde ser estudada.

Na tabela, cada linha representa um serviço em um trimestre. Dividimos as finalizadas com duração válida pelo total de solicitações daquela linha. Por exemplo, 118 dos 1.192 pedidos de avaliação de obstáculo na calçada do terceiro trimestre de 2026 atenderam aos critérios. O resultado, aproximadamente 9,9%, mede a disponibilidade do desfecho para o estudo.

```python
saida = Path('dados_polieus')
saida.mkdir(exist_ok=True)
df['elegivel'] = df['Status'].eq('Finalizada') & df['dias'].ge(0)
cobertura = df.groupby(['periodo', 'Serviço']).agg(
    solicitacoes=('Serviço','size'), finalizadas_com_duracao=('elegivel','sum'))
cobertura['percentual_elegivel'] = (100 * cobertura.finalizadas_com_duracao / cobertura.solicitacoes).round(1)
print(cobertura.to_string())
cobertura.to_csv(saida / '01_cobertura_por_servico_e_trimestre.csv', sep=';', decimal=',', encoding='utf-8-sig')
reg = df.loc[df.elegivel, ['periodo','Serviço','Distrito','abertura','finalizacao','dias']].copy()
reg.columns = ['periodo_fonte','servico','distrito','abertura','finalizacao','duracao_dias']
reg.to_csv(saida / '03_solicitacoes_com_duracao_em_dias.csv', index=False)
print('Solicitações elegíveis:', len(reg))
```

A tabela exportada conserva a data de finalização para conferir a duração e saber quando o desfecho se tornou conhecido. Durante o treinamento do modelo, usar essa data para prever a duração na abertura revelaria parte da resposta. As entradas deverão ser escolhidas entre as informações disponíveis quando a solicitação é registrada. O cabeçalho dos arquivos públicos também limita a verificação de duplicidade: solicitações com o mesmo serviço, distrito e dia podem corresponder a ocorrências distintas.

## 6. Como transformar solicitações em perfis territoriais?

A unidade de análise passa a ser um distrito em um mês de abertura. Para cada combinação observada, contamos quantos registros pertencem a cada serviço e dividimos essas contagens pelo volume total do perfil. Em um exemplo hipotético, cinco registros de manutenção de calçada pública entre vinte chamados correspondem a uma participação de 25%. As oito participações devem somar 100%.

A coluna de volume ajuda a avaliar a estabilidade dessas proporções. Um único chamado altera muito mais um perfil com dois registros do que um perfil com duzentos. Os registros sem distrito ficam fora da agregação territorial. A tabela apresenta somente combinações observadas, pois a ausência de registros pode ter mais de uma explicação.

```python
observados = df.dropna(subset=['Distrito','mes'])
contagens = observados.groupby(['Distrito','mes','Serviço']).size().unstack(fill_value=0).reindex(columns=servicos, fill_value=0)
volume = contagens.sum(axis=1)
perfis = contagens.div(volume, axis=0).add_prefix('proporcao_')
# As proporções de cada distrito-mês devem somar 100%.
assert (perfis.sum(axis=1).sub(1).abs() < 1e-9).all()
assert volume.sum() == len(observados)
perfis.insert(0,'total_solicitacoes',volume)
perfis.rename_axis(index=['distrito','mes']).to_csv(saida / '02_perfis_por_distrito_e_mes.csv')
print('Perfis observados:', len(perfis), '| perfis com menos de cinco registros:', int((volume < 5).sum()))
print('Volume por perfil (quartis):')
print(volume.describe().round(1).to_string())
print('Composição do recorte territorial (%):')
print((100 * contagens.sum() / volume.sum()).round(1).to_string())
```

## 7. Como interpretar o gráfico de cobertura?

Escolha um serviço no eixo vertical e compare as duas barras dele. A cor indica o trimestre de abertura. O comprimento mostra a porcentagem de solicitações com status Finalizada e duração válida. A tabela da seção 5 fornece os números absolutos usados para calcular cada barra.

```python
import matplotlib.pyplot as plt
grafico = cobertura['percentual_elegivel'].unstack('periodo')
ax = grafico.plot.barh(figsize=(11, 5), color=['#22577a', '#df862b'])
ax.set(xlabel='Registros elegíveis para regressão (%)', ylabel='', xlim=(0, 100),
       title='Cobertura de duração observada por serviço e trimestre de abertura')
ax.legend(title='Trimestre')
plt.tight_layout()
plt.show()
```

Nos pedidos de manutenção de calçada pública, a cobertura passa de 84,9% no trimestre de 2025 para 59,1% no trimestre de 2026. Isso significa que, no arquivo mais recente, uma fração menor dos registros desse serviço forneceu duração para a regressão. Para investigar se o atendimento ficou mais rápido ou mais lento, seria preciso analisar as durações e levar em conta o acompanhamento dos casos ainda abertos.

As aberturas analisadas pertencem a dois trimestres separados por seis meses sem observação neste recorte. A comparação descreve as edições selecionadas. Uma análise de tendência ao longo de todo o intervalo exigiria incluir e verificar os trimestres intermediários.

Depois de gravar as tabelas, o notebook abre cada CSV novamente e compara o total de linhas lidas com o total da tabela preparada em memória. Essa conferência verifica se todas as linhas esperadas chegaram ao arquivo. A pasta de saída chama-se `dados_polieus`.
